```python

# ================================================================
# QUESTION 1: DECISION TREE - IRIS FLOWER CLASSIFICATION
# Complete AI Lab Assignment with Improved Visualizations
# ================================================================

# ======================== IMPORT LIBRARIES =======================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from matplotlib.ticker import PercentFormatter

# ======================== VISUALIZATION SETTINGS =================

sns.set_theme(style="whitegrid", context="notebook")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11
plt.rcParams["axes.titleweight"] = "bold"

# ======================== PART 1 ================================
# LOAD AND ANALYZE IRIS DATASET
# ================================================================

file_path = r"C:\CE\CE-7\GitHub\AI LAB\AILAB7\IRIS.csv"

data = pd.read_csv(file_path)

print("=" * 70)
print("PART 1: IRIS DATASET LOADED SUCCESSFULLY")
print("=" * 70)

print("\nFirst Five Rows:")
print(data.head())

print("\nDataset Shape:", data.shape)

print("\nDataset Information:")
data.info()

print("\nStatistical Description:")
print(data.describe())

# Actual dataset column names
feature_names = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

target_column = "species"

# Validate dataset columns
required_columns = feature_names + [target_column]

if not all(col in data.columns for col in required_columns):
    raise ValueError(
        "Expected columns were not found. Available columns: "
        + str(data.columns.tolist())
    )

# ======================== PART 2 ================================
# DATA QUALITY ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("PART 2: DATA QUALITY ANALYSIS")
print("=" * 70)

missing_values = data.isnull().sum()
total_missing = missing_values.sum()
total_cells = data.size

print("\nMissing Values:")
print(missing_values)

print("\nTotal Missing Values:", total_missing)
print("Duplicate Rows:", data.duplicated().sum())

# Missing-value percentage
missing_percentage = (missing_values / len(data)) * 100

# ======================== VISUALIZATION 1 ========================
# DATASET OVERVIEW DASHBOARD
# ================================================================

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

overview_labels = [
    "Total Samples",
    "Input Features",
    "Target Variable"
]

overview_values = [
    len(data),
    len(feature_names),
    1
]

bars = axes[0].bar(
    overview_labels,
    overview_values,
    color=["#2563EB", "#14B8A6", "#F59E0B"]
)

axes[0].set_title("Dataset Overview")
axes[0].set_ylabel("Count")
axes[0].tick_params(axis="x", rotation=15)

for bar in bars:
    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 2,
        int(bar.get_height()),
        ha="center",
        fontweight="bold"
    )

# Dataset completeness
complete_percentage = (
    (total_cells - total_missing) / total_cells
) * 100

axes[1].pie(
    [complete_percentage, 100 - complete_percentage],
    labels=["Complete", "Missing"],
    autopct="%1.1f%%",
    startangle=90,
    colors=["#10B981", "#EF4444"],
    wedgeprops={"width": 0.4, "edgecolor": "white"}
)

axes[1].set_title("Dataset Completeness")

# Feature types
axes[2].bar(
    ["Numerical", "Categorical"],
    [len(feature_names), 1],
    color=["#6366F1", "#F97316"]
)

axes[2].set_title("Feature Types")
axes[2].set_ylabel("Number of Columns")

plt.suptitle(
    "IRIS DATASET - ANALYTICS DASHBOARD",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ======================== VISUALIZATION 2 ========================
# CLEAR MISSING VALUE VISUALIZATION
# ================================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    gridspec_kw={"width_ratios": [1.5, 1]}
)

# Left: Missing value counts
colors = [
    "#EF4444" if value > 0 else "#10B981"
    for value in missing_values
]

bars = axes[0].barh(
    missing_values.index,
    missing_values.values,
    color=colors,
    edgecolor="white"
)

axes[0].set_title(
    "Missing Values by Column",
    fontsize=14,
    fontweight="bold"
)

axes[0].set_xlabel("Number of Missing Values")
axes[0].set_ylabel("Dataset Columns")

axes[0].set_xlim(
    0,
    max(1, int(missing_values.max()) + 2)
)

for bar, value in zip(bars, missing_values.values):
    axes[0].text(
        value + 0.05,
        bar.get_y() + bar.get_height() / 2,
        str(value),
        va="center",
        fontweight="bold"
    )

# Right: Dataset quality status
if total_missing == 0:

    axes[1].pie(
        [100],
        labels=["Complete Data"],
        colors=["#10B981"],
        autopct="%1.0f%%",
        startangle=90,
        wedgeprops={"width": 0.35, "edgecolor": "white"},
        textprops={"fontsize": 11}
    )

    axes[1].text(
        0, 0,
        "100%\nCOMPLETE",
        ha="center",
        va="center",
        fontsize=16,
        fontweight="bold",
        color="#047857"
    )

    axes[1].set_title("Dataset Quality Status")

else:

    axes[1].pie(
        [total_missing, total_cells - total_missing],
        labels=["Missing", "Available"],
        colors=["#EF4444", "#10B981"],
        autopct="%1.2f%%",
        startangle=90
    )

    axes[1].set_title("Missing vs Available Data")

plt.suptitle(
    "MISSING VALUE ANALYSIS",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# Print clear interpretation
if total_missing == 0:
    print("\nRESULT: No missing values found.")
    print("All dataset columns are 100% complete.")
else:
    print("\nRESULT: Missing values were detected.")
    print("Missing values will be handled before model training.")

# ======================== DATA CLEANING ==========================

# Remove duplicate records
data = data.drop_duplicates().copy()

# Remove rows with missing values
data = data.dropna().copy()

print("\nDataset Shape After Cleaning:", data.shape)

# ======================== PART 3 ================================
# IDENTIFY INPUT FEATURES AND TARGET
# ================================================================

X = data[feature_names]
y = data[target_column]

print("\n" + "=" * 70)
print("PART 3: INPUT FEATURES AND TARGET IDENTIFIED")
print("=" * 70)

print("\nInput Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())

print("\nInput Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nUnique Species:")
print(y.unique())

# ======================== VISUALIZATION 3 ========================
# SPECIES DISTRIBUTION
# ================================================================

plt.figure(figsize=(10, 6))

species_counts = y.value_counts()

bars = plt.bar(
    species_counts.index,
    species_counts.values,
    color=["#6366F1", "#14B8A6", "#F59E0B"],
    edgecolor="white",
    width=0.6
)

plt.title(
    "IRIS FLOWER SPECIES DISTRIBUTION",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Flower Species")
plt.ylabel("Number of Samples")

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 1,
        int(bar.get_height()),
        ha="center",
        fontweight="bold"
    )

plt.show()

# ======================== VISUALIZATION 4 ========================
# FEATURE DISTRIBUTION WITH HISTOGRAM AND KDE
# ================================================================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

feature_colors = [
    "#6366F1",
    "#14B8A6",
    "#F59E0B",
    "#EC4899"
]

for i, feature in enumerate(feature_names):

    ax = axes[i // 2, i % 2]

    sns.histplot(
        data=data,
        x=feature,
        hue="species",
        kde=True,
        element="step",
        common_norm=False,
        ax=ax
    )

    ax.set_title(
        feature.replace("_", " ").title(),
        fontweight="bold"
    )

plt.suptitle(
    "FEATURE DISTRIBUTION ANALYSIS",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ======================== VISUALIZATION 5 ========================
# PAIR PLOT
# ================================================================

pair_plot = sns.pairplot(
    data,
    vars=feature_names,
    hue="species",
    diag_kind="kde",
    corner=True,
    plot_kws={"alpha": 0.75, "s": 45}
)

pair_plot.fig.suptitle(
    "IRIS FLOWER - FEATURE RELATIONSHIPS",
    y=1.02,
    fontsize=16,
    fontweight="bold"
)

plt.show()

# ======================== VISUALIZATION 6 ========================
# CORRELATION HEATMAP
# ================================================================

plt.figure(figsize=(9, 7))

correlation = data[feature_names].corr()

mask = np.triu(
    np.ones_like(correlation, dtype=bool)
)

sns.heatmap(
    correlation,
    mask=mask,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=2,
    square=True,
    vmin=-1,
    vmax=1,
    cbar_kws={"label": "Correlation"}
)

plt.title(
    "FEATURE CORRELATION MATRIX",
    fontsize=15,
    fontweight="bold"
)

plt.show()

# ======================== PART 4 ================================
# TRAINING AND TESTING DATA SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 70)
print("PART 4: TRAINING AND TESTING DATA SPLIT")
print("=" * 70)

print("\nTotal Samples:", len(X))
print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))

# ======================== VISUALIZATION 7 ========================
# TRAIN TEST DONUT CHART
# ================================================================

plt.figure(figsize=(8, 7))

plt.pie(
    [len(X_train), len(X_test)],
    labels=["Training Data (80%)", "Testing Data (20%)"],
    autopct="%1.1f%%",
    startangle=90,
    colors=["#6366F1", "#F59E0B"],
    wedgeprops={"width": 0.4, "edgecolor": "white"},
    textprops={"fontsize": 12}
)

plt.text(
    0, 0,
    f"{len(X)}\nSamples",
    ha="center",
    va="center",
    fontsize=16,
    fontweight="bold"
)

plt.title(
    "TRAINING VS TESTING DATA",
    fontsize=15,
    fontweight="bold"
)

plt.show()

# ======================== PART 5 ================================
# TRAIN DECISION TREE MODEL
# ================================================================

model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

model.fit(X_train, y_train)

print("\n" + "=" * 70)
print("PART 5: DECISION TREE MODEL TRAINED SUCCESSFULLY")
print("=" * 70)

print(model)

# ======================== VISUALIZATION 8 ========================
# DECISION TREE STRUCTURE
# ================================================================

plt.figure(figsize=(22, 12))

plot_tree(
    model,
    feature_names=feature_names,
    class_names=[str(c) for c in model.classes_],
    filled=True,
    rounded=True,
    proportion=True,
    impurity=True,
    fontsize=9
)

plt.title(
    "DECISION TREE CLASSIFICATION STRUCTURE",
    fontsize=16,
    fontweight="bold"
)

plt.show()

# ======================== PART 6 ================================
# GENERATE PREDICTIONS
# ================================================================

y_pred = model.predict(X_test)

print("\n" + "=" * 70)
print("PART 6: PREDICTIONS GENERATED SUCCESSFULLY")
print("=" * 70)

prediction_results = pd.DataFrame({
    "Actual Species": y_test.values,
    "Predicted Species": y_pred,
    "Correct Prediction": y_test.values == y_pred
})

print("\nPrediction Results:")
print(prediction_results.to_string(index=False))

# ======================== VISUALIZATION 9 ========================
# ACTUAL VS PREDICTED
# ================================================================

class_labels = list(model.classes_)

actual_numeric = [
    class_labels.index(value) for value in y_test
]

predicted_numeric = [
    class_labels.index(value) for value in y_pred
]

sample_numbers = np.arange(len(y_test))

plt.figure(figsize=(14, 6))

plt.scatter(
    sample_numbers,
    actual_numeric,
    label="Actual",
    marker="o",
    s=80,
    alpha=0.8
)

plt.scatter(
    sample_numbers,
    predicted_numeric,
    label="Predicted",
    marker="x",
    s=80,
    alpha=0.8
)

plt.title(
    "ACTUAL VS PREDICTED FLOWER SPECIES",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Testing Sample Number")
plt.ylabel("Flower Species")

plt.yticks(
    range(len(class_labels)),
    class_labels
)

plt.legend()
plt.grid(alpha=0.3)
plt.show()

# ======================== PART 7 ================================
# MODEL PERFORMANCE EVALUATION
# ================================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 70)
print("PART 7: MODEL PERFORMANCE EVALUATION")
print("=" * 70)

print(f"\nAccuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

# ======================== VISUALIZATION 10 =======================
# PERFORMANCE METRICS DASHBOARD
# ================================================================

metric_names = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

metric_values = [
    accuracy,
    precision,
    recall,
    f1
]

fig, axes = plt.subplots(1, 4, figsize=(16, 5))

metric_colors = [
    "#6366F1",
    "#14B8A6",
    "#F59E0B",
    "#EC4899"
]

for ax, name, value, color in zip(
    axes,
    metric_names,
    metric_values,
    metric_colors
):

    ax.barh(
        ["Score"],
        [value],
        color=color,
        height=0.4
    )

    ax.set_xlim(0, 1)
    ax.set_title(name, fontweight="bold")

    ax.text(
        value / 2,
        0,
        f"{value * 100:.1f}%",
        ha="center",
        va="center",
        color="white",
        fontweight="bold",
        fontsize=12
    )

    ax.set_xlabel("Performance")

plt.suptitle(
    "MODEL PERFORMANCE DASHBOARD",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ======================== VISUALIZATION 11 =======================
# CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=model.classes_
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=2,
    linecolor="white",
    xticklabels=model.classes_,
    yticklabels=model.classes_,
    annot_kws={"size": 16, "weight": "bold"}
)

plt.title(
    "CONFUSION MATRIX - IRIS CLASSIFICATION",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")

plt.show()

print("\nConfusion Matrix:")
print(cm)

# ======================== PART 8 ================================
# FEATURE IMPORTANCE
# ================================================================

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=True
)

print("\n" + "=" * 70)
print("PART 8: FEATURE IMPORTANCE")
print("=" * 70)

print(feature_importance.sort_values(
    by="Importance",
    ascending=False
).to_string(index=False))

# ======================== VISUALIZATION 12 =======================
# FEATURE IMPORTANCE RANKING
# ================================================================

plt.figure(figsize=(10, 6))

bars = plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"],
    color=["#F59E0B", "#EC4899", "#14B8A6", "#6366F1"],
    edgecolor="white"
)

plt.title(
    "FEATURE IMPORTANCE - DECISION TREE",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Importance Score")
plt.ylabel("Input Features")

for bar in bars:

    value = bar.get_width()

    plt.text(
        value + 0.01,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.3f}",
        va="center",
        fontweight="bold"
    )

plt.xlim(
    0,
    max(0.1, feature_importance["Importance"].max() + 0.15)
)

plt.show()

# ======================== PART 9 ================================
# FINAL CONCLUSION
# ================================================================

print("\n" + "=" * 70)
print("FINAL CONCLUSION")
print("=" * 70)

print(f"""
The Decision Tree Classifier was successfully implemented
using the Iris Flower Dataset.

DATASET INFORMATION:
- Total Samples: {len(data)}
- Input Features: 4
- Target Variable: Species
- Missing Values: {total_missing}

INPUT FEATURES:
1. Sepal Length
2. Sepal Width
3. Petal Length
4. Petal Width

TARGET CLASSES:
{", ".join(map(str, model.classes_))}

MODEL CONFIGURATION:
- Algorithm: Decision Tree Classifier
- Criterion: Gini
- Training Data: 80%
- Testing Data: 20%

MODEL PERFORMANCE:
- Accuracy: {accuracy * 100:.2f}%
- Precision: {precision * 100:.2f}%
- Recall: {recall * 100:.2f}%
- F1 Score: {f1 * 100:.2f}%

The model learned classification rules from the training data
and predicted the species of the testing samples.

Petal measurements are generally important for distinguishing
Iris flower species. The feature importance graph identifies
which measurements were most useful in this trained model.

The confusion matrix shows correct and incorrect predictions.

The model was evaluated using accuracy, precision, recall,
F1 score, and confusion matrix.

Therefore, the Decision Tree algorithm successfully performs
Iris flower classification using four numerical measurements.
""")

print("=" * 70)
print("QUESTION 1 COMPLETED SUCCESSFULLY")
print("=" * 70)
```
