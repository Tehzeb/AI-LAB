```python

# ============================================================================
#                 AI LAB | MACHINE LEARNING ANALYTICS COMPLETE
#                        MACHINE LEARNING ASSIGNMENT
# ============================================================================

# ============================================================================
# PART 1: IMPORT LIBRARIES
# ============================================================================

import os
import glob
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, label_binarize
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc
)

warnings.filterwarnings("ignore")

# ============================================================================
# PART 2: CONFIGURATION AND VISUALIZATION THEME
# ============================================================================

DATA_FOLDER = r"C:\CE\CE-7\GitHub\AI LAB\AILAB7"

RANDOM_STATE = 42
TEST_SIZE = 0.20

plt.style.use("dark_background")

COLORS = {
    "cyan": "#00F2FE",
    "blue": "#3B82F6",
    "purple": "#A855F7",
    "green": "#10B981",
    "orange": "#F59E0B",
    "red": "#EF4444",
    "pink": "#EC4899",
    "background": "#0B0F19"
}

sns.set_theme(style="darkgrid")

def heading(title):
    print("\n" + "=" * 75)
    print(title.center(75))
    print("=" * 75)


def finish_plot():
    plt.tight_layout()
    plt.show()


heading("NEON AI LAB | MACHINE LEARNING ANALYTICS")

# ============================================================================
# PART 3: FIND DATASET FILES
# ============================================================================

def find_dataset(keyword):

    files = glob.glob(
        os.path.join(DATA_FOLDER, "*.csv")
    )

    matches = [
        file for file in files
        if keyword.lower() in os.path.basename(file).lower()
    ]

    if not matches:
        raise FileNotFoundError(
            f"{keyword}.csv was not found in {DATA_FOLDER}"
        )

    if len(matches) > 1:
        raise ValueError(
            f"Multiple files found for {keyword}: {matches}. "
            "Keep only one matching CSV file or select its exact path."
        )

    return matches[0]


breast_path = find_dataset("Breastcancer")
wine_path = find_dataset("Wine")
diabetes_path = find_dataset("Diabetes")

print("\nDATASET FILES FOUND:")

print(os.path.basename(breast_path))
print(os.path.basename(wine_path))
print(os.path.basename(diabetes_path))

# ============================================================================
# PART 4: LOAD DATASETS
# ============================================================================

breast_df = pd.read_csv(breast_path)
wine_df = pd.read_csv(wine_path)
diabetes_df = pd.read_csv(diabetes_path)

datasets = {
    "Breastcancer": breast_df,
    "Wine": wine_df,
    "Diabetes": diabetes_df
}

for name, df in datasets.items():

    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
    )

    print(f"\nLoaded successfully: {name}.csv")
    print("Shape:", df.shape)

# ============================================================================
# PART 5: DATASET TARGET CONFIGURATION
# ============================================================================

TARGETS = {
    "Breastcancer": "diagnosis",
    "Wine": "class",
    "Diabetes": "class"
}

for name, target in TARGETS.items():

    if target not in datasets[name].columns:
        raise KeyError(
            f"Target column '{target}' not found in {name} dataset."
        )

    print(f"{name}: target column = '{target}'")

# ============================================================================
# PART 6: DATASET ANALYSIS
# ============================================================================

heading("PART 6 | DATASET ANALYSIS")

for name, df in datasets.items():

    target = TARGETS[name]

    heading(f"DATASET: {name}")

    print("Dataset Shape:", df.shape)
    print("Target Column:", target)

    print("\nFirst Five Records:")
    display(df.head())

    print("\nData Types:")
    display(df.dtypes.to_frame("Data Type"))

    print("\nMissing Values:")
    display(df.isnull().sum().to_frame("Missing Values"))

    print("\nStatistical Summary:")
    display(df.describe(include="all").T)

    print("\nTarget Distribution:")
    display(
        df[target]
        .astype(str)
        .str.strip()
        .value_counts()
        .to_frame("Frequency")
    )

    # Visualization 1: Dataset dimensions
    plt.figure(figsize=(8, 4))

    plt.bar(
        ["Rows", "Columns"],
        [df.shape[0], df.shape[1]],
        color=[COLORS["cyan"], COLORS["purple"]]
    )

    plt.title(f"{name} - Dataset Dimensions")
    plt.ylabel("Count")

    finish_plot()

    # Visualization 2: Missing values
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    plt.figure(figsize=(10, 4))

    if len(missing) > 0:

        plt.bar(
            missing.index.astype(str),
            missing.values,
            color=COLORS["orange"]
        )

        plt.xticks(rotation=45, ha="right")

    else:

        plt.text(
            0.5, 0.5,
            "No Missing Values",
            ha="center",
            va="center",
            fontsize=18,
            color=COLORS["green"],
            transform=plt.gca().transAxes
        )

    plt.title(f"{name} - Missing Value Analysis")
    plt.ylabel("Missing Count")

    finish_plot()

    # Visualization 3: Target distribution
    target_counts = (
        df[target]
        .astype(str)
        .str.strip()
        .value_counts()
    )

    plt.figure(figsize=(8, 5))

    plt.bar(
        target_counts.index.astype(str),
        target_counts.values,
        color=COLORS["purple"]
    )

    plt.title(f"{name} - Target Class Distribution")
    plt.xlabel("Target Class")
    plt.ylabel("Number of Samples")

    finish_plot()

# ============================================================================
# PART 7: DATA CLEANING AND PREPROCESSING
# ============================================================================

heading("PART 7 | DATA CLEANING AND PREPROCESSING")

def prepare_dataset(df, target, name):

    df = df.copy()

    # Clean column names
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
    )

    # Normalize target values
    df[target] = df[target].astype("string").str.strip()

    # Handle missing target values
    df[target] = df[target].replace(
        ["", "nan", "None", "NaN"],
        pd.NA
    )

    df = df.dropna(subset=[target]).copy()

    # Remove duplicate records
    before = len(df)
    df = df.drop_duplicates()

    print(
        f"\n{name}: Duplicate rows removed:",
        before - len(df)
    )

    # Separate features and target
    y_original = df[target].astype(str).str.strip()

    X = df.drop(columns=[target]).copy()

    # Remove identifier columns
    id_columns = [
        col for col in X.columns
        if col in [
            "id",
            "index",
            "unnamed_0",
            "unnamed_32",
            "no_pation",
            "patient_id"
        ]
    ]

    X.drop(columns=id_columns, inplace=True, errors="ignore")

    # Remove completely empty columns
    empty_columns = [
        col for col in X.columns
        if X[col].isnull().all()
    ]

    X.drop(columns=empty_columns, inplace=True, errors="ignore")

    # Remove constant columns
    constant_columns = [
        col for col in X.columns
        if X[col].nunique(dropna=False) <= 1
    ]

    X.drop(columns=constant_columns, inplace=True, errors="ignore")

    # Encode target
    encoder = LabelEncoder()

    y = pd.Series(
        encoder.fit_transform(y_original),
        index=X.index,
        name=target
    )

    class_counts = y_original.value_counts()

    print("\nPREPROCESSING:", name)
    print("Samples:", len(X))
    print("Features:", X.shape[1])
    print("Target Classes:", encoder.classes_.tolist())

    print("\nCleaned Target Distribution:")
    display(class_counts.to_frame("Frequency"))

    print("Removed ID Columns:", id_columns)
    print("Removed Empty Columns:", empty_columns)
    print("Removed Constant Columns:", constant_columns)

    if class_counts.min() < 2:

        raise ValueError(
            f"{name}: A target class has fewer than two samples. "
            "Check the dataset labels."
        )

    return X, y, encoder


X_breast, y_breast, breast_encoder = prepare_dataset(
    breast_df, "diagnosis", "Breastcancer"
)

X_wine, y_wine, wine_encoder = prepare_dataset(
    wine_df, "class", "Wine"
)

X_diabetes, y_diabetes, diabetes_encoder = prepare_dataset(
    diabetes_df, "class", "Diabetes"
)

# ============================================================================
# PART 8: CLEANED DIABETES LABEL VISUALIZATION
# ============================================================================

heading("PART 8 | DIABETES LABEL CLEANING")

diabetes_labels = diabetes_encoder.inverse_transform(
    y_diabetes.to_numpy()
)

diabetes_counts = pd.Series(
    diabetes_labels
).value_counts()

print("Diabetes classes after whitespace cleaning:")
display(diabetes_counts.to_frame("Frequency"))

plt.figure(figsize=(9, 5))

bars = plt.bar(
    diabetes_counts.index.astype(str),
    diabetes_counts.values,
    color=COLORS["cyan"]
)

plt.title("Diabetes - Cleaned Class Distribution")
plt.xlabel("Diabetes Class")
plt.ylabel("Number of Samples")

for bar in bars:

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )

finish_plot()

# ============================================================================
# PART 9: FEATURE DISTRIBUTION VISUALIZATION
# ============================================================================

heading("PART 9 | FEATURE DISTRIBUTION ANALYSIS")

prepared_datasets = {
    "Breastcancer": (X_breast, y_breast),
    "Wine": (X_wine, y_wine),
    "Diabetes": (X_diabetes, y_diabetes)
}

for name, (X, y) in prepared_datasets.items():

    # Select numeric features
    numeric_columns = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    if not numeric_columns:
        continue

    # Boxplot of first numeric features
    selected = numeric_columns[:8]

    plt.figure(figsize=(12, 5))

    sns.boxplot(
        data=X[selected],
        palette="husl"
    )

    plt.title(f"{name} - Feature Distribution")
    plt.xticks(rotation=45, ha="right")

    finish_plot()

    # Correlation heatmap
    correlation = X[numeric_columns].corr()

    plt.figure(figsize=(11, 8))

    sns.heatmap(
        correlation,
        cmap="coolwarm",
        annot=False,
        linewidths=0.3
    )

    plt.title(f"{name} - Feature Correlation Heatmap")

    finish_plot()

# ============================================================================
# PART 10: TRAIN TEST SPLITTING
# ============================================================================

heading("PART 10 | TRAIN TEST SPLITTING")

def split_dataset(X, y, name):

    counts = y.value_counts()

    if counts.min() < 2:
        raise ValueError(
            f"{name}: Cannot stratify because a class has fewer than "
            "two observations."
        )

    # Ensure test set has enough observations for all classes
    number_of_classes = y.nunique()

    test_count = max(
        int(np.ceil(len(y) * TEST_SIZE)),
        number_of_classes
    )

    if test_count >= len(y):
        raise ValueError(
            f"{name}: Not enough samples for train/test splitting."
        )

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_count,
        random_state=RANDOM_STATE,
        stratify=y
    )

    print(f"\n{name}")
    print("Training Samples:", len(X_train))
    print("Testing Samples:", len(X_test))

    # Visualization
    plt.figure(figsize=(7, 4))

    plt.bar(
        ["Training", "Testing"],
        [len(X_train), len(X_test)],
        color=[COLORS["green"], COLORS["orange"]]
    )

    plt.title(f"{name} - Train Test Split")
    plt.ylabel("Number of Samples")

    finish_plot()

    return X_train, X_test, y_train, y_test


# ============================================================================
# PART 11: PREPROCESSING PIPELINE
# ============================================================================

def create_preprocessor(X):

    numeric_features = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_features = X.select_dtypes(
        exclude=np.number
    ).columns.tolist()

    transformers = []

    if numeric_features:

        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            )
        ])

        transformers.append(
            ("numeric", numeric_pipeline, numeric_features)
        )

    if categorical_features:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "encoder",
                OneHotEncoder(handle_unknown="ignore")
            )
        ])

        transformers.append(
            ("categorical", categorical_pipeline, categorical_features)
        )

    return ColumnTransformer(transformers)


# ============================================================================
# PART 12: MODEL CONFIGURATION
# ============================================================================

heading("PART 12 | MACHINE LEARNING MODEL CONFIGURATION")

model_configs = {

    "Breastcancer": {
        "model": DecisionTreeClassifier(
            max_depth=5,
            min_samples_split=4,
            random_state=RANDOM_STATE
        ),
        "dataset": (X_breast, y_breast),
        "encoder": breast_encoder,
        "description": "Decision Tree Classifier"
    },

    "Wine": {
        "model": RandomForestClassifier(
            n_estimators=200,
            max_depth=8,
            random_state=RANDOM_STATE,
            n_jobs=-1
        ),
        "dataset": (X_wine, y_wine),
        "encoder": wine_encoder,
        "description": "Random Forest Classifier"
    },

    "Diabetes": {
        "model": GradientBoostingClassifier(
            n_estimators=150,
            learning_rate=0.05,
            max_depth=2,
            random_state=RANDOM_STATE
        ),
        "dataset": (X_diabetes, y_diabetes),
        "encoder": diabetes_encoder,
        "description": "Gradient Boosting Classifier"
    }

}

for name, config in model_configs.items():

    print("\nDataset:", name)
    print("Algorithm:", config["description"])

# ============================================================================
# PART 13: MODEL TRAINING AND EVALUATION
# ============================================================================

heading("PART 13 | MODEL TRAINING AND EVALUATION")

results = {}

def train_model(name, config):

    X, y = config["dataset"]

    X_train, X_test, y_train, y_test = split_dataset(
        X, y, name
    )

    preprocessor = create_preprocessor(X)

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", config["model"])
    ])

    print("\nTraining:", config["description"])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    probabilities = pipeline.predict_proba(X_test)

    encoder = config["encoder"]

    original_classes = encoder.classes_

    # Calculate metrics
    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    print("\nMODEL PERFORMANCE")
    print("-" * 50)

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1 Score:  {f1:.4f}")

    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            predictions,
            labels=np.arange(len(original_classes)),
            target_names=[str(c) for c in original_classes],
            zero_division=0
        )
    )

    # Store results
    results[name] = {

        "model": pipeline,
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test,
        "predictions": predictions,
        "probabilities": probabilities,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "classes": original_classes,
        "feature_names": None
    }

    # Confusion matrix
    cm = confusion_matrix(
        y_test,
        predictions,
        labels=np.arange(len(original_classes))
    )

    plt.figure(figsize=(7, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="mako",
        xticklabels=original_classes,
        yticklabels=original_classes
    )

    plt.title(f"{name} - Confusion Matrix")
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")

    finish_plot()

    # Metric visualization
    metric_names = [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]

    metric_values = [
        accuracy,
        precision,
        recall,
        f1
    ]

    plt.figure(figsize=(8, 5))

    bars = plt.bar(
        metric_names,
        metric_values,
        color=[
            COLORS["cyan"],
            COLORS["purple"],
            COLORS["green"],
            COLORS["orange"]
        ]
    )

    plt.ylim(0, 1.1)
    plt.title(f"{name} - Model Evaluation Metrics")
    plt.ylabel("Score")

    for bar in bars:

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.02,
            f"{bar.get_height():.3f}",
            ha="center"
        )

    finish_plot()

    return pipeline


for name, config in model_configs.items():

    train_model(name, config)

# ============================================================================
# PART 14: FEATURE IMPORTANCE
# ============================================================================

heading("PART 14 | FEATURE IMPORTANCE ANALYSIS")

for name, result in results.items():

    pipeline = result["model"]

    preprocessor = pipeline.named_steps["preprocessor"]
    model = pipeline.named_steps["model"]

    feature_names = preprocessor.get_feature_names_out()

    result["feature_names"] = feature_names

    importance = model.feature_importances_

    importance_df = pd.DataFrame({

        "Feature": feature_names,
        "Importance": importance

    }).sort_values(
        by="Importance",
        ascending=False
    )

    print(f"\n{name} - Top 10 Important Features")

    display(importance_df.head(10))

    plt.figure(figsize=(10, 6))

    top_features = importance_df.head(10).sort_values(
        "Importance"
    )

    plt.barh(
        top_features["Feature"],
        top_features["Importance"],
        color=COLORS["cyan"]
    )

    plt.title(f"{name} - Top 10 Feature Importance")
    plt.xlabel("Importance")

    finish_plot()

# ============================================================================
# PART 15: DECISION TREE STRUCTURE
# ============================================================================

heading("PART 15 | DECISION TREE STRUCTURE")

tree_model = results["Breastcancer"]["model"].named_steps["model"]

tree_features = results["Breastcancer"]["feature_names"]

plt.figure(figsize=(24, 12))

plot_tree(
    tree_model,
    feature_names=tree_features,
    class_names=[
        str(c) for c in breast_encoder.classes_
    ],
    filled=True,
    rounded=True,
    fontsize=7
)

plt.title("Breast Cancer - Decision Tree Structure")

finish_plot()

# ============================================================================
# PART 16: ROC CURVE ANALYSIS
# ============================================================================

heading("PART 16 | ROC CURVE ANALYSIS")

for name, result in results.items():

    y_test = result["y_test"].to_numpy()

    probabilities = result["probabilities"]

    classes = np.unique(y_test)

    plt.figure(figsize=(8, 6))

    if len(classes) == 2:

        fpr, tpr, thresholds = roc_curve(
            y_test,
            probabilities[:, 1]
        )

        roc_auc = auc(fpr, tpr)

        plt.plot(
            fpr,
            tpr,
            linewidth=2,
            label=f"AUC = {roc_auc:.3f}"
        )

    else:

        y_binary = label_binarize(
            y_test,
            classes=np.arange(len(result["classes"]))
        )

        for i in range(len(result["classes"])):

            if y_binary[:, i].sum() == 0:
                continue

            fpr, tpr, thresholds = roc_curve(
                y_binary[:, i],
                probabilities[:, i]
            )

            roc_auc = auc(fpr, tpr)

            plt.plot(
                fpr,
                tpr,
                linewidth=2,
                label=f"Class {result['classes'][i]} AUC = {roc_auc:.3f}"
            )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        color=COLORS["red"],
        label="Random Classifier"
    )

    plt.title(f"{name} - ROC Curve")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend()

    finish_plot()

# ============================================================================
# PART 17: ACTUAL VS PREDICTED CLASS DISTRIBUTION
# ============================================================================

heading("PART 17 | ACTUAL VS PREDICTED COMPARISON")

for name, result in results.items():

    encoder = model_configs[name]["encoder"]

    actual = encoder.inverse_transform(
        result["y_test"].to_numpy()
    )

    predicted = encoder.inverse_transform(
        result["predictions"]
    )

    comparison = pd.DataFrame({

        "Actual": actual,
        "Predicted": predicted

    })

    print(f"\n{name} - Actual vs Predicted")
    display(comparison.head(15))

    actual_counts = pd.Series(actual).value_counts()
    predicted_counts = pd.Series(predicted).value_counts()

    all_classes = encoder.classes_

    x = np.arange(len(all_classes))
    width = 0.35

    plt.figure(figsize=(9, 5))

    plt.bar(
        x - width / 2,
        [actual_counts.get(c, 0) for c in all_classes],
        width,
        label="Actual",
        color=COLORS["cyan"]
    )

    plt.bar(
        x + width / 2,
        [predicted_counts.get(c, 0) for c in all_classes],
        width,
        label="Predicted",
        color=COLORS["purple"]
    )

    plt.xticks(x, all_classes)
    plt.title(f"{name} - Actual vs Predicted Distribution")
    plt.xlabel("Class")
    plt.ylabel("Number of Samples")
    plt.legend()

    finish_plot()

# ============================================================================
# PART 18: OVERALL MODEL COMPARISON
# ============================================================================

heading("PART 18 | OVERALL MODEL COMPARISON")

comparison_df = pd.DataFrame({

    name: {

        "Accuracy": result["accuracy"],
        "Precision": result["precision"],
        "Recall": result["recall"],
        "F1 Score": result["f1"]

    }

    for name, result in results.items()

}).T

print("\nFINAL MODEL PERFORMANCE COMPARISON")

display(comparison_df.round(4))

# Comparison bar chart
comparison_df.plot(
    kind="bar",
    figsize=(12, 6),
    ylim=(0, 1.1),
    width=0.75
)

plt.title("Overall Machine Learning Model Comparison")
plt.xlabel("Dataset and Algorithm")
plt.ylabel("Performance Score")
plt.xticks(rotation=0)
plt.legend(loc="lower right")

finish_plot()

# ============================================================================
# PART 19: RADAR CHART
# ============================================================================

heading("PART 19 | MODEL PERFORMANCE RADAR CHART")

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

angles = np.linspace(
    0,
    2 * np.pi,
    len(metrics),
    endpoint=False
).tolist()

angles += angles[:1]

plt.figure(figsize=(8, 8))

ax = plt.subplot(111, polar=True)

for name, result in results.items():

    values = [
        result["accuracy"],
        result["precision"],
        result["recall"],
        result["f1"]
    ]

    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=name
    )

    ax.fill(
        angles,
        values,
        alpha=0.10
    )

ax.set_xticks(angles[:-1])
ax.set_xticklabels(metrics)

ax.set_ylim(0, 1)

plt.title("Machine Learning Performance Radar")
plt.legend(loc="upper right", bbox_to_anchor=(1.35, 1.1))

finish_plot()

# ============================================================================
# PART 20: MODEL RANKING
# ============================================================================

heading("PART 20 | MODEL PERFORMANCE RANKING")

ranking = comparison_df.sort_values(
    by="Accuracy",
    ascending=False
)

display(ranking.round(4))

plt.figure(figsize=(9, 5))

bars = plt.barh(
    ranking.index,
    ranking["Accuracy"],
    color=[
        COLORS["green"],
        COLORS["cyan"],
        COLORS["purple"]
    ][:len(ranking)]
)

plt.xlabel("Accuracy")
plt.title("Model Accuracy Ranking")
plt.xlim(0, 1.1)

for bar in bars:

    plt.text(
        bar.get_width() + 0.01,
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():.3f}",
        va="center"
    )

finish_plot()

# ============================================================================
# PART 21: MODEL STRENGTHS AND LIMITATIONS
# ============================================================================

heading("PART 21 | MODEL STRENGTHS AND LIMITATIONS")

model_information = {

    "Decision Tree": {

        "strengths": [
            "Easy to understand",
            "Simple interpretation",
            "Handles nonlinear patterns"
        ],

        "limitations": [
            "Can overfit",
            "Sensitive to data changes",
            "May be less stable"
        ]

    },

    "Random Forest": {

        "strengths": [
            "Reduces overfitting",
            "Uses multiple trees",
            "Good general performance"
        ],

        "limitations": [
            "More computational cost",
            "Less interpretable",
            "Requires more memory"
        ]

    },

    "Gradient Boosting": {

        "strengths": [
            "Learns complex patterns",
            "Improves weak learners",
            "Strong predictive ability"
        ],

        "limitations": [
            "Sensitive to parameters",
            "Can require longer training",
            "May overfit without tuning"
        ]

    }

}

for model_name, information in model_information.items():

    print("\n", model_name)

    print("\nStrengths:")

    for item in information["strengths"]:
        print(" +", item)

    print("\nLimitations:")

    for item in information["limitations"]:
        print(" -", item)

    # Visualization
    fig, ax = plt.subplots(figsize=(10, 5))

    strength_count = len(information["strengths"])
    limitation_count = len(information["limitations"])

    ax.barh(
        information["strengths"],
        [1] * strength_count,
        label="Strengths",
        color=COLORS["green"]
    )

    ax.barh(
        information["limitations"],
        [-1] * limitation_count,
        label="Limitations",
        color=COLORS["red"]
    )

    ax.axvline(0, color="white", linewidth=1)

    ax.set_title(f"{model_name} - Strengths vs Limitations")
    ax.set_xlabel("Positive / Negative Characteristics")
    ax.legend()

    finish_plot()

# ============================================================================
# PART 22: EXPLANATION OF EVALUATION METRICS
# ============================================================================

heading("PART 22 | EVALUATION METRICS EXPLANATION")

metric_explanations = {

    "Accuracy":
        "Percentage of all predictions that are correct.",

    "Precision":
        "Percentage of predicted positive cases that are actually positive.",

    "Recall":
        "Percentage of actual positive cases correctly identified.",

    "F1 Score":
        "Harmonic mean of precision and recall."

}

for metric, explanation in metric_explanations.items():

    print(f"\n{metric}:")
    print(explanation)

# Visualization
plt.figure(figsize=(11, 5))

plt.bar(
    metric_explanations.keys(),
    [1, 1, 1, 1],
    color=[
        COLORS["cyan"],
        COLORS["purple"],
        COLORS["green"],
        COLORS["orange"]
    ]
)

plt.title("Understanding Classification Evaluation Metrics")
plt.ylabel("Metric Importance (Conceptual)")
plt.ylim(0, 1.3)

finish_plot()

# ============================================================================
# PART 23: FINAL RESULTS
# ============================================================================

heading("PART 23 | FINAL RESULTS")

print("\nDATASETS AND MODELS:")

print("1. Breast Cancer -> Decision Tree Classifier")
print("2. Wine -> Random Forest Classifier")
print("3. Diabetes -> Gradient Boosting Classifier")

print("\nFINAL PERFORMANCE:")

display(comparison_df.round(4))

best_model = comparison_df["Accuracy"].idxmax()

print("\nBest accuracy among the three tested models:")
print(best_model)

print(
    "Accuracy:",
    f"{comparison_df.loc[best_model, 'Accuracy']:.4f}"
)

print("\nAll datasets processed successfully.")
print("All three models trained and evaluated.")

heading("NEON AI LAB | EXECUTION COMPLETED")

# ============================================================================
# END OF MACHINE LEARNING ASSIGNMENT
# ============================================================================
```

```python
# ============================================================================
#                 NEON AI LAB | MACHINE LEARNING ANALYTICS
#                 COMPLETE MACHINE LEARNING ASSIGNMENT
# ============================================================================

# ============================================================================
# PART 1: IMPORT LIBRARIES
# ============================================================================

import os
import glob
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, label_binarize
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc
)

warnings.filterwarnings("ignore")

# ============================================================================
# PART 2: CONFIGURATION AND VISUALIZATION THEME
# ============================================================================

DATA_FOLDER = r"C:\CE\CE-7\GitHub\AI LAB\AILAB7"

RANDOM_STATE = 42
TEST_SIZE = 0.20

plt.style.use("dark_background")

COLORS = {
    "cyan": "#00F2FE",
    "blue": "#3B82F6",
    "purple": "#A855F7",
    "green": "#10B981",
    "orange": "#F59E0B",
    "red": "#EF4444",
    "pink": "#EC4899",
    "background": "#0B0F19"
}

sns.set_theme(style="darkgrid")

def heading(title):
    print("\n" + "=" * 75)
    print(title.center(75))
    print("=" * 75)


def finish_plot():
    plt.tight_layout()
    plt.show()


heading("NEON AI LAB | MACHINE LEARNING ANALYTICS")

# ============================================================================
# PART 3: FIND DATASET FILES
# ============================================================================

def find_dataset(keyword):

    files = glob.glob(
        os.path.join(DATA_FOLDER, "*.csv")
    )

    matches = [
        file for file in files
        if keyword.lower() in os.path.basename(file).lower()
    ]

    if not matches:
        raise FileNotFoundError(
            f"{keyword}.csv was not found in {DATA_FOLDER}"
        )

    if len(matches) > 1:
        raise ValueError(
            f"Multiple files found for {keyword}: {matches}. "
            "Keep only one matching CSV file or select its exact path."
        )

    return matches[0]


breast_path = find_dataset("Breastcancer")
wine_path = find_dataset("Wine")
diabetes_path = find_dataset("Diabetes")

print("\nDATASET FILES FOUND:")

print(os.path.basename(breast_path))
print(os.path.basename(wine_path))
print(os.path.basename(diabetes_path))

# ============================================================================
# PART 4: LOAD DATASETS
# ============================================================================

breast_df = pd.read_csv(breast_path)
wine_df = pd.read_csv(wine_path)
diabetes_df = pd.read_csv(diabetes_path)

datasets = {
    "Breastcancer": breast_df,
    "Wine": wine_df,
    "Diabetes": diabetes_df
}

for name, df in datasets.items():

    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
    )

    print(f"\nLoaded successfully: {name}.csv")
    print("Shape:", df.shape)

# ============================================================================
# PART 5: DATASET TARGET CONFIGURATION
# ============================================================================

TARGETS = {
    "Breastcancer": "diagnosis",
    "Wine": "class",
    "Diabetes": "class"
}

for name, target in TARGETS.items():

    if target not in datasets[name].columns:
        raise KeyError(
            f"Target column '{target}' not found in {name} dataset."
        )

    print(f"{name}: target column = '{target}'")

# ============================================================================
# PART 6: DATASET ANALYSIS
# ============================================================================

heading("PART 6 | DATASET ANALYSIS")

for name, df in datasets.items():

    target = TARGETS[name]

    heading(f"DATASET: {name}")

    print("Dataset Shape:", df.shape)
    print("Target Column:", target)

    print("\nFirst Five Records:")
    display(df.head())

    print("\nData Types:")
    display(df.dtypes.to_frame("Data Type"))

    print("\nMissing Values:")
    display(df.isnull().sum().to_frame("Missing Values"))

    print("\nStatistical Summary:")
    display(df.describe(include="all").T)

    print("\nTarget Distribution:")
    display(
        df[target]
        .astype(str)
        .str.strip()
        .value_counts()
        .to_frame("Frequency")
    )

    # Visualization 1: Dataset dimensions
    plt.figure(figsize=(8, 4))

    plt.bar(
        ["Rows", "Columns"],
        [df.shape[0], df.shape[1]],
        color=[COLORS["cyan"], COLORS["purple"]]
    )

    plt.title(f"{name} - Dataset Dimensions")
    plt.ylabel("Count")

    finish_plot()

    # Visualization 2: Missing values
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    plt.figure(figsize=(10, 4))

    if len(missing) > 0:

        plt.bar(
            missing.index.astype(str),
            missing.values,
            color=COLORS["orange"]
        )

        plt.xticks(rotation=45, ha="right")

    else:

        plt.text(
            0.5, 0.5,
            "No Missing Values",
            ha="center",
            va="center",
            fontsize=18,
            color=COLORS["green"],
            transform=plt.gca().transAxes
        )

    plt.title(f"{name} - Missing Value Analysis")
    plt.ylabel("Missing Count")

    finish_plot()

    # Visualization 3: Target distribution
    target_counts = (
        df[target]
        .astype(str)
        .str.strip()
        .value_counts()
    )

    plt.figure(figsize=(8, 5))

    plt.bar(
        target_counts.index.astype(str),
        target_counts.values,
        color=COLORS["purple"]
    )

    plt.title(f"{name} - Target Class Distribution")
    plt.xlabel("Target Class")
    plt.ylabel("Number of Samples")

    finish_plot()

# ============================================================================
# PART 7: DATA CLEANING AND PREPROCESSING
# ============================================================================

heading("PART 7 | DATA CLEANING AND PREPROCESSING")

def prepare_dataset(df, target, name):

    df = df.copy()

    # Clean column names
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
    )

    # Normalize target values
    df[target] = df[target].astype("string").str.strip()

    # Handle missing target values
    df[target] = df[target].replace(
        ["", "nan", "None", "NaN"],
        pd.NA
    )

    df = df.dropna(subset=[target]).copy()

    # Remove duplicate records
    before = len(df)
    df = df.drop_duplicates()

    print(
        f"\n{name}: Duplicate rows removed:",
        before - len(df)
    )

    # Separate features and target
    y_original = df[target].astype(str).str.strip()

    X = df.drop(columns=[target]).copy()

    # Remove identifier columns
    id_columns = [
        col for col in X.columns
        if col in [
            "id",
            "index",
            "unnamed_0",
            "unnamed_32",
            "no_pation",
            "patient_id"
        ]
    ]

    X.drop(columns=id_columns, inplace=True, errors="ignore")

    # Remove completely empty columns
    empty_columns = [
        col for col in X.columns
        if X[col].isnull().all()
    ]

    X.drop(columns=empty_columns, inplace=True, errors="ignore")

    # Remove constant columns
    constant_columns = [
        col for col in X.columns
        if X[col].nunique(dropna=False) <= 1
    ]

    X.drop(columns=constant_columns, inplace=True, errors="ignore")

    # Encode target
    encoder = LabelEncoder()

    y = pd.Series(
        encoder.fit_transform(y_original),
        index=X.index,
        name=target
    )

    class_counts = y_original.value_counts()

    print("\nPREPROCESSING:", name)
    print("Samples:", len(X))
    print("Features:", X.shape[1])
    print("Target Classes:", encoder.classes_.tolist())

    print("\nCleaned Target Distribution:")
    display(class_counts.to_frame("Frequency"))

    print("Removed ID Columns:", id_columns)
    print("Removed Empty Columns:", empty_columns)
    print("Removed Constant Columns:", constant_columns)

    if class_counts.min() < 2:

        raise ValueError(
            f"{name}: A target class has fewer than two samples. "
            "Check the dataset labels."
        )

    return X, y, encoder


X_breast, y_breast, breast_encoder = prepare_dataset(
    breast_df, "diagnosis", "Breastcancer"
)

X_wine, y_wine, wine_encoder = prepare_dataset(
    wine_df, "class", "Wine"
)

X_diabetes, y_diabetes, diabetes_encoder = prepare_dataset(
    diabetes_df, "class", "Diabetes"
)

# ============================================================================
# PART 8: CLEANED DIABETES LABEL VISUALIZATION
# ============================================================================

heading("PART 8 | DIABETES LABEL CLEANING")

diabetes_labels = diabetes_encoder.inverse_transform(
    y_diabetes.to_numpy()
)

diabetes_counts = pd.Series(
    diabetes_labels
).value_counts()

print("Diabetes classes after whitespace cleaning:")
display(diabetes_counts.to_frame("Frequency"))

plt.figure(figsize=(9, 5))

bars = plt.bar(
    diabetes_counts.index.astype(str),
    diabetes_counts.values,
    color=COLORS["cyan"]
)

plt.title("Diabetes - Cleaned Class Distribution")
plt.xlabel("Diabetes Class")
plt.ylabel("Number of Samples")

for bar in bars:

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )

finish_plot()

# ============================================================================
# PART 9: FEATURE DISTRIBUTION VISUALIZATION
# ============================================================================

heading("PART 9 | FEATURE DISTRIBUTION ANALYSIS")

prepared_datasets = {
    "Breastcancer": (X_breast, y_breast),
    "Wine": (X_wine, y_wine),
    "Diabetes": (X_diabetes, y_diabetes)
}

for name, (X, y) in prepared_datasets.items():

    # Select numeric features
    numeric_columns = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    if not numeric_columns:
        continue

    # Boxplot of first numeric features
    selected = numeric_columns[:8]

    plt.figure(figsize=(12, 5))

    sns.boxplot(
        data=X[selected],
        palette="husl"
    )

    plt.title(f"{name} - Feature Distribution")
    plt.xticks(rotation=45, ha="right")

    finish_plot()

    # Correlation heatmap
    correlation = X[numeric_columns].corr()

    plt.figure(figsize=(11, 8))

    sns.heatmap(
        correlation,
        cmap="coolwarm",
        annot=False,
        linewidths=0.3
    )

    plt.title(f"{name} - Feature Correlation Heatmap")

    finish_plot()

# ============================================================================
# PART 10: TRAIN TEST SPLITTING
# ============================================================================

heading("PART 10 | TRAIN TEST SPLITTING")

def split_dataset(X, y, name):

    counts = y.value_counts()

    if counts.min() < 2:
        raise ValueError(
            f"{name}: Cannot stratify because a class has fewer than "
            "two observations."
        )

    # Ensure test set has enough observations for all classes
    number_of_classes = y.nunique()

    test_count = max(
        int(np.ceil(len(y) * TEST_SIZE)),
        number_of_classes
    )

    if test_count >= len(y):
        raise ValueError(
            f"{name}: Not enough samples for train/test splitting."
        )

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_count,
        random_state=RANDOM_STATE,
        stratify=y
    )

    print(f"\n{name}")
    print("Training Samples:", len(X_train))
    print("Testing Samples:", len(X_test))

    # Visualization
    plt.figure(figsize=(7, 4))

    plt.bar(
        ["Training", "Testing"],
        [len(X_train), len(X_test)],
        color=[COLORS["green"], COLORS["orange"]]
    )

    plt.title(f"{name} - Train Test Split")
    plt.ylabel("Number of Samples")

    finish_plot()

    return X_train, X_test, y_train, y_test


# ============================================================================
# PART 11: PREPROCESSING PIPELINE
# ============================================================================

def create_preprocessor(X):

    numeric_features = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_features = X.select_dtypes(
        exclude=np.number
    ).columns.tolist()

    transformers = []

    if numeric_features:

        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            )
        ])

        transformers.append(
            ("numeric", numeric_pipeline, numeric_features)
        )

    if categorical_features:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="most_frequent")
            ),
            (
                "encoder",
                OneHotEncoder(handle_unknown="ignore")
            )
        ])

        transformers.append(
            ("categorical", categorical_pipeline, categorical_features)
        )

    return ColumnTransformer(transformers)


# ============================================================================
# PART 12: MODEL CONFIGURATION
# ============================================================================

heading("PART 12 | MACHINE LEARNING MODEL CONFIGURATION")

model_configs = {

    "Breastcancer": {
        "model": DecisionTreeClassifier(
            max_depth=5,
            min_samples_split=4,
            random_state=RANDOM_STATE
        ),
        "dataset": (X_breast, y_breast),
        "encoder": breast_encoder,
        "description": "Decision Tree Classifier"
    },

    "Wine": {
        "model": RandomForestClassifier(
            n_estimators=200,
            max_depth=8,
            random_state=RANDOM_STATE,
            n_jobs=-1
        ),
        "dataset": (X_wine, y_wine),
        "encoder": wine_encoder,
        "description": "Random Forest Classifier"
    },

    "Diabetes": {
        "model": GradientBoostingClassifier(
            n_estimators=150,
            learning_rate=0.05,
            max_depth=2,
            random_state=RANDOM_STATE
        ),
        "dataset": (X_diabetes, y_diabetes),
        "encoder": diabetes_encoder,
        "description": "Gradient Boosting Classifier"
    }

}

for name, config in model_configs.items():

    print("\nDataset:", name)
    print("Algorithm:", config["description"])

# ============================================================================
# PART 13: MODEL TRAINING AND EVALUATION
# ============================================================================

heading("PART 13 | MODEL TRAINING AND EVALUATION")

results = {}

def train_model(name, config):

    X, y = config["dataset"]

    X_train, X_test, y_train, y_test = split_dataset(
        X, y, name
    )

    preprocessor = create_preprocessor(X)

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", config["model"])
    ])

    print("\nTraining:", config["description"])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    probabilities = pipeline.predict_proba(X_test)

    encoder = config["encoder"]

    original_classes = encoder.classes_

    # Calculate metrics
    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    print("\nMODEL PERFORMANCE")
    print("-" * 50)

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1 Score:  {f1:.4f}")

    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            predictions,
            labels=np.arange(len(original_classes)),
            target_names=[str(c) for c in original_classes],
            zero_division=0
        )
    )

    # Store results
    results[name] = {

        "model": pipeline,
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test,
        "predictions": predictions,
        "probabilities": probabilities,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "classes": original_classes,
        "feature_names": None
    }

    # Confusion matrix
    cm = confusion_matrix(
        y_test,
        predictions,
        labels=np.arange(len(original_classes))
    )

    plt.figure(figsize=(7, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="mako",
        xticklabels=original_classes,
        yticklabels=original_classes
    )

    plt.title(f"{name} - Confusion Matrix")
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")

    finish_plot()

    # Metric visualization
    metric_names = [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]

    metric_values = [
        accuracy,
        precision,
        recall,
        f1
    ]

    plt.figure(figsize=(8, 5))

    bars = plt.bar(
        metric_names,
        metric_values,
        color=[
            COLORS["cyan"],
            COLORS["purple"],
            COLORS["green"],
            COLORS["orange"]
        ]
    )

    plt.ylim(0, 1.1)
    plt.title(f"{name} - Model Evaluation Metrics")
    plt.ylabel("Score")

    for bar in bars:

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.02,
            f"{bar.get_height():.3f}",
            ha="center"
        )

    finish_plot()

    return pipeline


for name, config in model_configs.items():

    train_model(name, config)

# ============================================================================
# PART 14: FEATURE IMPORTANCE
# ============================================================================

heading("PART 14 | FEATURE IMPORTANCE ANALYSIS")

for name, result in results.items():

    pipeline = result["model"]

    preprocessor = pipeline.named_steps["preprocessor"]
    model = pipeline.named_steps["model"]

    feature_names = preprocessor.get_feature_names_out()

    result["feature_names"] = feature_names

    importance = model.feature_importances_

    importance_df = pd.DataFrame({

        "Feature": feature_names,
        "Importance": importance

    }).sort_values(
        by="Importance",
        ascending=False
    )

    print(f"\n{name} - Top 10 Important Features")

    display(importance_df.head(10))

    plt.figure(figsize=(10, 6))

    top_features = importance_df.head(10).sort_values(
        "Importance"
    )

    plt.barh(
        top_features["Feature"],
        top_features["Importance"],
        color=COLORS["cyan"]
    )

    plt.title(f"{name} - Top 10 Feature Importance")
    plt.xlabel("Importance")

    finish_plot()

# ============================================================================
# PART 15: DECISION TREE STRUCTURE
# ============================================================================

heading("PART 15 | DECISION TREE STRUCTURE")

tree_model = results["Breastcancer"]["model"].named_steps["model"]

tree_features = results["Breastcancer"]["feature_names"]

plt.figure(figsize=(24, 12))

plot_tree(
    tree_model,
    feature_names=tree_features,
    class_names=[
        str(c) for c in breast_encoder.classes_
    ],
    filled=True,
    rounded=True,
    fontsize=7
)

plt.title("Breast Cancer - Decision Tree Structure")

finish_plot()

# ============================================================================
# PART 16: ROC CURVE ANALYSIS
# ============================================================================

heading("PART 16 | ROC CURVE ANALYSIS")

for name, result in results.items():

    y_test = result["y_test"].to_numpy()

    probabilities = result["probabilities"]

    classes = np.unique(y_test)

    plt.figure(figsize=(8, 6))

    if len(classes) == 2:

        fpr, tpr, thresholds = roc_curve(
            y_test,
            probabilities[:, 1]
        )

        roc_auc = auc(fpr, tpr)

        plt.plot(
            fpr,
            tpr,
            linewidth=2,
            label=f"AUC = {roc_auc:.3f}"
        )

    else:

        y_binary = label_binarize(
            y_test,
            classes=np.arange(len(result["classes"]))
        )

        for i in range(len(result["classes"])):

            if y_binary[:, i].sum() == 0:
                continue

            fpr, tpr, thresholds = roc_curve(
                y_binary[:, i],
                probabilities[:, i]
            )

            roc_auc = auc(fpr, tpr)

            plt.plot(
                fpr,
                tpr,
                linewidth=2,
                label=f"Class {result['classes'][i]} AUC = {roc_auc:.3f}"
            )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        color=COLORS["red"],
        label="Random Classifier"
    )

    plt.title(f"{name} - ROC Curve")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend()

    finish_plot()

# ============================================================================
# PART 17: ACTUAL VS PREDICTED CLASS DISTRIBUTION
# ============================================================================

heading("PART 17 | ACTUAL VS PREDICTED COMPARISON")

for name, result in results.items():

    encoder = model_configs[name]["encoder"]

    actual = encoder.inverse_transform(
        result["y_test"].to_numpy()
    )

    predicted = encoder.inverse_transform(
        result["predictions"]
    )

    comparison = pd.DataFrame({

        "Actual": actual,
        "Predicted": predicted

    })

    print(f"\n{name} - Actual vs Predicted")
    display(comparison.head(15))

    actual_counts = pd.Series(actual).value_counts()
    predicted_counts = pd.Series(predicted).value_counts()

    all_classes = encoder.classes_

    x = np.arange(len(all_classes))
    width = 0.35

    plt.figure(figsize=(9, 5))

    plt.bar(
        x - width / 2,
        [actual_counts.get(c, 0) for c in all_classes],
        width,
        label="Actual",
        color=COLORS["cyan"]
    )

    plt.bar(
        x + width / 2,
        [predicted_counts.get(c, 0) for c in all_classes],
        width,
        label="Predicted",
        color=COLORS["purple"]
    )

    plt.xticks(x, all_classes)
    plt.title(f"{name} - Actual vs Predicted Distribution")
    plt.xlabel("Class")
    plt.ylabel("Number of Samples")
    plt.legend()

    finish_plot()

# ============================================================================
# PART 18: OVERALL MODEL COMPARISON
# ============================================================================

heading("PART 18 | OVERALL MODEL COMPARISON")

comparison_df = pd.DataFrame({

    name: {

        "Accuracy": result["accuracy"],
        "Precision": result["precision"],
        "Recall": result["recall"],
        "F1 Score": result["f1"]

    }

    for name, result in results.items()

}).T

print("\nFINAL MODEL PERFORMANCE COMPARISON")

display(comparison_df.round(4))

# Comparison bar chart
comparison_df.plot(
    kind="bar",
    figsize=(12, 6),
    ylim=(0, 1.1),
    width=0.75
)

plt.title("Overall Machine Learning Model Comparison")
plt.xlabel("Dataset and Algorithm")
plt.ylabel("Performance Score")
plt.xticks(rotation=0)
plt.legend(loc="lower right")

finish_plot()

# ============================================================================
# PART 19: RADAR CHART
# ============================================================================

heading("PART 19 | MODEL PERFORMANCE RADAR CHART")

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

angles = np.linspace(
    0,
    2 * np.pi,
    len(metrics),
    endpoint=False
).tolist()

angles += angles[:1]

plt.figure(figsize=(8, 8))

ax = plt.subplot(111, polar=True)

for name, result in results.items():

    values = [
        result["accuracy"],
        result["precision"],
        result["recall"],
        result["f1"]
    ]

    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=name
    )

    ax.fill(
        angles,
        values,
        alpha=0.10
    )

ax.set_xticks(angles[:-1])
ax.set_xticklabels(metrics)

ax.set_ylim(0, 1)

plt.title("Machine Learning Performance Radar")
plt.legend(loc="upper right", bbox_to_anchor=(1.35, 1.1))

finish_plot()

# ============================================================================
# PART 20: MODEL RANKING
# ============================================================================

heading("PART 20 | MODEL PERFORMANCE RANKING")

ranking = comparison_df.sort_values(
    by="Accuracy",
    ascending=False
)

display(ranking.round(4))

plt.figure(figsize=(9, 5))

bars = plt.barh(
    ranking.index,
    ranking["Accuracy"],
    color=[
        COLORS["green"],
        COLORS["cyan"],
        COLORS["purple"]
    ][:len(ranking)]
)

plt.xlabel("Accuracy")
plt.title("Model Accuracy Ranking")
plt.xlim(0, 1.1)

for bar in bars:

    plt.text(
        bar.get_width() + 0.01,
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():.3f}",
        va="center"
    )

finish_plot()

# ============================================================================
# PART 21: MODEL STRENGTHS AND LIMITATIONS
# ============================================================================

heading("PART 21 | MODEL STRENGTHS AND LIMITATIONS")

model_information = {

    "Decision Tree": {

        "strengths": [
            "Easy to understand",
            "Simple interpretation",
            "Handles nonlinear patterns"
        ],

        "limitations": [
            "Can overfit",
            "Sensitive to data changes",
            "May be less stable"
        ]

    },

    "Random Forest": {

        "strengths": [
            "Reduces overfitting",
            "Uses multiple trees",
            "Good general performance"
        ],

        "limitations": [
            "More computational cost",
            "Less interpretable",
            "Requires more memory"
        ]

    },

    "Gradient Boosting": {

        "strengths": [
            "Learns complex patterns",
            "Improves weak learners",
            "Strong predictive ability"
        ],

        "limitations": [
            "Sensitive to parameters",
            "Can require longer training",
            "May overfit without tuning"
        ]

    }

}

for model_name, information in model_information.items():

    print("\n", model_name)

    print("\nStrengths:")

    for item in information["strengths"]:
        print(" +", item)

    print("\nLimitations:")

    for item in information["limitations"]:
        print(" -", item)

    # Visualization
    fig, ax = plt.subplots(figsize=(10, 5))

    strength_count = len(information["strengths"])
    limitation_count = len(information["limitations"])

    ax.barh(
        information["strengths"],
        [1] * strength_count,
        label="Strengths",
        color=COLORS["green"]
    )

    ax.barh(
        information["limitations"],
        [-1] * limitation_count,
        label="Limitations",
        color=COLORS["red"]
    )

    ax.axvline(0, color="white", linewidth=1)

    ax.set_title(f"{model_name} - Strengths vs Limitations")
    ax.set_xlabel("Positive / Negative Characteristics")
    ax.legend()

    finish_plot()

# ============================================================================
# PART 22: EXPLANATION OF EVALUATION METRICS
# ============================================================================

heading("PART 22 | EVALUATION METRICS EXPLANATION")

metric_explanations = {

    "Accuracy":
        "Percentage of all predictions that are correct.",

    "Precision":
        "Percentage of predicted positive cases that are actually positive.",

    "Recall":
        "Percentage of actual positive cases correctly identified.",

    "F1 Score":
        "Harmonic mean of precision and recall."

}

for metric, explanation in metric_explanations.items():

    print(f"\n{metric}:")
    print(explanation)

# Visualization
plt.figure(figsize=(11, 5))

plt.bar(
    metric_explanations.keys(),
    [1, 1, 1, 1],
    color=[
        COLORS["cyan"],
        COLORS["purple"],
        COLORS["green"],
        COLORS["orange"]
    ]
)

plt.title("Understanding Classification Evaluation Metrics")
plt.ylabel("Metric Importance (Conceptual)")
plt.ylim(0, 1.3)

finish_plot()

# ============================================================================
# PART 23: FINAL RESULTS
# ============================================================================

heading("PART 23 | FINAL RESULTS")

print("\nDATASETS AND MODELS:")

print("1. Breast Cancer -> Decision Tree Classifier")
print("2. Wine -> Random Forest Classifier")
print("3. Diabetes -> Gradient Boosting Classifier")

print("\nFINAL PERFORMANCE:")

display(comparison_df.round(4))

best_model = comparison_df["Accuracy"].idxmax()

print("\nBest accuracy among the three tested models:")
print(best_model)

print(
    "Accuracy:",
    f"{comparison_df.loc[best_model, 'Accuracy']:.4f}"
)

print("\nAll datasets processed successfully.")
print("All three models trained and evaluated.")

heading("EXECUTION COMPLETED")
```
